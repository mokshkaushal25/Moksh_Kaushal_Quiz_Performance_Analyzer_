## 6. Your individual project challenge
**Challenge:** Calculate average quiz score; identify top scorer and most consistent student using standard deviation.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# 6. Individual Challenge - Using FOR LOOP
import numpy as np

quiz_cols = ['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']

averages = []
std_devs = []

# Har bacche ke liye loop
for index, row in df.iterrows():
    marks = []
    for col in quiz_cols:
        marks.append(row[col])
    
    avg = sum(marks) / len(marks)
    std = np.std(marks)
    
    averages.append(avg)
    std_devs.append(std)

df['Average'] = averages
df['Std_Dev'] = std_devs

print(df[['Student_ID', 'Name', 'Average', 'Std_Dev']].head())

# Top Scorer dhoondna - Jiska Average sabse zyada
top_index = 0
for i in range(len(df)):
    if df['Average'].iloc[i] > df['Average'].iloc[top_index]:
        top_index = i

print("\nTop Scorer:")
print(df.loc[top_index, ['Student_ID', 'Name', 'Average']])

# Most Consistent dhoondna - Jiska Std_Dev sabse KAM
consistent_index = 0
for i in range(len(df)):
    if df['Std_Dev'].iloc[i] < df['Std_Dev'].iloc[consistent_index]:
        consistent_index = i

print("\nMost Consistent Student:")
print(df.loc[consistent_index, ['Student_ID', 'Name', 'Average', 'Std_Dev']])
```
